```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
y = df.posted_rate

print(f"skewness  {y.skew():8.3f}")
print(f"kurtosis  {y.kurtosis():8.3f}")
print(f"mean      {y.mean():8.2f}")
print(f"median    {y.median():8.2f}")
for q in [0.5, 0.9, 0.99, 0.999, 1.0]:
    print(f"q{q:<7} {y.quantile(q):8.2f}")
print(f"\nlog skewness  {np.log(y).skew():.3f}")
print(f"log mean      {np.log(y).mean():.3f}")
print(f"log median    {np.log(y).median():.3f}")

fig, ax = plt.subplots(1, 2, figsize=(11, 4))

ax[0].hist(y, bins=80, color="#4477aa")
ax[0].set_title(f"posted_rate   (skew {y.skew():.2f})")
ax[0].set_xlabel("dollars")

ax[1].hist(np.log(y), bins=80, color="#cc6677")
ax[1].set_title(f"log(posted_rate)   (skew {np.log(y).skew():.2f})")
ax[1].set_xlabel("log dollars")

for a in ax:
    a.set_ylabel("loads")
fig.tight_layout()
```
