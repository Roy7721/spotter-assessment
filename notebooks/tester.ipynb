```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))
print("ROOT =", ROOT)

import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

from data import load_all
from train import add_features, time_split, gbm, FEATURES

train, valid, dec = load_all()
fit, hold = time_split(train)
fit_f, hold_f = add_features(fit), add_features(hold)

m = gbm().fit(fit_f[FEATURES], np.log(fit_f.posted_rate))
pred = np.exp(m.predict(hold_f[FEATURES]))
ratio = hold_f.posted_rate.values / pred
bad = (ratio > 1.5) | (ratio < 0.7)
print(f"contaminated {bad.sum()} of {len(bad)} ({100*bad.mean():.2f}%)")

X  = StandardScaler().fit_transform(hold_f[FEATURES])
Xp = StandardScaler().fit_transform(
    np.c_[hold_f[FEATURES].values, np.log(hold_f.posted_rate.values)])

fig, ax = plt.subplots(1, 2, figsize=(12, 5))
for a, d, title in [(ax[0], X, "features only"),
                    (ax[1], Xp, "features + log(price)")]:
    p = PCA(n_components=2).fit_transform(d)
    a.scatter(p[~bad, 0], p[~bad, 1], s=4, c="#cccccc", label="normal")
    a.scatter(p[bad, 0], p[bad, 1], s=18, c="#cc3311", label="contaminated")
    a.set_title(f"PCA — {title}")
    a.set_xlabel("PC1"); a.set_ylabel("PC2"); a.legend()
fig.tight_layout()
fig.savefig("../reports/pca_no_separation.png", dpi=150)
```

```python
# evidence: is a column a property of the DAY, or of the LOAD?
# between-day sd = how much the days differ from each other
# within-day sd  = how much loads on the SAME day differ from each other
for col in ["market_index", "quote_signal"]:
    g = train.groupby("date")[col]
    between = g.mean().std()
    within  = g.std().mean()
    verdict = "property of the DAY" if between > within else "property of the LOAD"
    print(f"{col:14} between-day sd {between:.4f}   within-day sd {within:.4f}   -> {verdict}")

# the same thing visible on a single day
day = train[train.date == "2025-03-15"]
print(f"\n{len(day)} loads on 2025-03-15, first 10:")
print(day[["pickup", "delivery", "market_index", "quote_signal"]].head(10).to_string(index=False))
print(f"\n  market_index spread that day : {day.market_index.max() - day.market_index.min():.5f}")
print(f"  quote_signal spread that day : {day.quote_signal.max() - day.quote_signal.min():.5f}")
```

```python
# no posted_rate in validation.csv -> cannot score. so: are the outputs SANE?
pred = pd.read_csv(ROOT / "validation_predictions.csv")
decp = pd.read_csv(ROOT / "december-predictions.csv")

# 1. LEVEL check: does this lane really trade at the price i predicted?
lane = train[(train.pickup == decp.pickup[0]) & (train.delivery == decp.delivery[0])]
print(f"lane  {decp.pickup[0]} -> {decp.delivery[0]}")
print(f"  same lane in training : {len(lane)} loads   "
      f"${lane.posted_rate.min():.2f} - ${lane.posted_rate.max():.2f}   "
      f"median ${lane.posted_rate.median():.2f}")
print(f"  my december chart     : ${decp.predicted_rate.min():.2f} - ${decp.predicted_rate.max():.2f}")
```

```python
# 2. DISTRIBUTION check: do the 12,000 predictions sit where real rates sit?
print("\nquantile    predicted    training")
for q in [0.01, 0.25, 0.50, 0.75, 0.99]:
    print(f"{q:8.0%} {pred.predicted_rate.quantile(q):12.2f} {train.posted_rate.quantile(q):11.2f}")
```

```python
# 3. SHAPE check: what weekly pattern did the model find on its own?
decp["date"] = pd.to_datetime(decp.date)
print("\ndecember mean by day of week:")
print(decp.groupby(decp.date.dt.day_name()).predicted_rate.mean().round(2).to_string())
```
